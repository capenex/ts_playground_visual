### 🎓 06 · 静态类型诊断与二维矩阵算法实战 (`Type Checker` + `2D Matrix`)
本课程展示如何将 **编译期静态类型防御** 与 **运行期算法可视化调试** 结合：
- **实时类型诊断**：当代码出现类型不匹配时，编辑器行尾会立即提示红色 `✗ TS` 诊断徽章。
- **双指针与二维网格**：变量名 `l` / `r` 会自动悬挂在一维数组下方；`dp: number[][]` 会自动渲染为二维矩阵表格。

```typescript
// 1. 静态类型防错实验室（可尝试取消倒数第二行注释，体验行尾实时红色 TS 诊断徽章）
interface RiskRule {
  readonly baseScore: number;
  multiplier: number;
}

function evaluateRisk(rule: RiskRule, factor: number): number {
  return rule.baseScore + rule.multiplier * factor;
}

const prodRule: RiskRule = { baseScore: 50, multiplier: 9 };
const safeScore: number = evaluateRisk(prodRule, 4);
// 试一试：取消下一行注释将触发静态类型错误（string 不能赋给 number）
// const brokenScore: number = "type_mismatch";
console.log("风控评估合规得分:", safeScore);
```

```typescript
// 2. 点击右上角 🐞 单步调试：观察一维数组下方自动悬挂的 [l, r] 双指针标签
const nums: number[] = [4, 9, 2, 11, 6, 8, 5];
let l: number = 0;
let r: number = nums.length - 1;
let windowBest: number = 0;

while (l < r) {
  const pairVal: number = nums[l] + nums[r];
  if (pairVal > windowBest) {
    windowBest = pairVal;
  }
  checkpoint(l, r, pairVal, windowBest);
  l = l + 1;
  r = r - 1;
}
console.log("双指针最大配对和:", windowBest);
```

```typescript
// 3. 点击右上角 🐞 单步调试：在右侧 WATCH 面板观察二维矩阵 dp: number[][] 的逐格填充
const rows: number = 3;
const cols: number = 4;
const dp: number[][] = [
  [1, 1, 1, 1],
  [1, 0, 0, 0],
  [1, 0, 0, 0]
];

for (let i: number = 1; i < rows; i++) {
  for (let j: number = 1; j < cols; j++) {
    dp[i][j] = dp[i - 1][j] + dp[i][j - 1];
    checkpoint(i, j, dp[i][j]);
  }
}

console.log("网格右下角不同路径总数:", dp[rows - 1][cols - 1]);
```
